# Domain-Specific Q&A Bot — RAG & Embedding Exploration Notebook
This notebook demonstrates loading documents, chunking text, generating vector embeddings using Sentence-Transformers, and querying ChromaDB.

```python
import sys
sys.path.append('..')

from sentence_transformers import SentenceTransformer
import chromadb
from app.retrieval import retrieve_context
from app.ingestion import chunk_text, clean_text
```

```python
# 1. Test Text Extraction & Chunking
sample_text = """
Retrieval-Augmented Generation (RAG) combines dense vector retrieval with LLMs.
Sentence-Transformers compute 384-dimensional embeddings for MiniLM-L6-v2.
ChromaDB persists vector collections on local disk.
"""
cleaned = clean_text(sample_text)
chunks = chunk_text(cleaned, chunk_size=100, chunk_overlap=20)
print(f"Generated {len(chunks)} chunks:")
for i, chunk in enumerate(chunks):
    print(f"  Chunk {i}: {chunk}")
```

```python
# 2. Test Embedding Generation
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
embeddings = model.encode(chunks)
print(f"Embeddings shape: {embeddings.shape}")
```

```python
# 3. Test Retrieval Service
results = retrieve_context("What is RAG?", top_k=2)
print("Retrieval Results:", results)
```
